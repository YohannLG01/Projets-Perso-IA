```python
# Fonction pour concaténer et diviser les dataset

def concat(train,test):
    return pd.concat([train, test],sort=True, axis=0, ignore_index=False)

def split(all_data):
    return all_data.loc[:890], all_data.loc[891:].drop(['Survived'], axis=1)
```
